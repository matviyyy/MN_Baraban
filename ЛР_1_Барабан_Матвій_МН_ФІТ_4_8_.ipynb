## Завдання 2. Скалярний добуток і довжина векторів

Дано вектори:
```python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])
```

1. Обчисліть скалярний добуток двома способами: `np.dot()` і оператором `@`.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами.

````python
u = np.array([2, -3, 6])
v = np.array([5, 1, 4])

dot1 = np.dot(u, v)
dot2 = u @ v
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
cos_angle = dot1 / (norm_u * norm_v)

print("np.dot(u, v) =", dot1)
print("u @ v =", dot2)
print("||u|| =", norm_u)
print("||v|| =", norm_v)
print("cos(angle) =", cos_angle)
````

## Завдання 15. Комутатор двох матриць

Для кожної пари матриць обчисліть комутатор
\[
[A,B] = AB - BA.
\]

Перша пара:
```python
A1 = np.array([[3, 0],
               [0, 7]])
B1 = np.array([[2, 0],
               [0, 5]])
```

Друга пара:
```python
A2 = np.array([[0, 2],
               [-1, 0]])
B2 = np.array([[4, 0],
               [0, 1]])
```

Визначте, для якої пари виконується `AB = BA`. Для перевірки використайте `np.allclose()`.

````python
A1 = np.array([[3, 0],
                [0, 7]])
B1 = np.array([[2, 0],
               [0, 5]])

A2 = np.array([[0, 2],
               [-1, 0]])
B2 = np.array([[4, 0],
               [0, 1]])

comm1 = A1 @ B1 - B1 @ A1
comm2 = A2 @ B2 - B2 @ A2

print("[A1, B1] =\n", comm1)
print("[A2, B2] =\n", comm2)
print("Для першої пари AB = BA:", np.allclose(A1 @ B1, B1 @ A1))
print("Для другої пари AB = BA:", np.allclose(A2 @ B2, B2 @ A2))
````

```python

    [2.0, 3.0]
])
```

1. Знайдіть власні значення та власні вектори.
2. Перевірте співвідношення `A @ V = V * λ`.
3. Виведіть результат перевірки через `np.allclose()`.

Використайте `np.linalg.eigh()`.

````python
A = np.array([
    [6.0, 2.0],
    [2.0, 3.0]
])

eigenvalues, V = np.linalg.eigh(A)
left = A @ V
right = V * eigenvalues

print("Власні значення:", eigenvalues)
print("Власні вектори (стовпці V):\n", V)
print("A @ V =\n", left)
print("V * λ =\n", right)
print("Перевірка A @ V = V * λ:", np.allclose(left, right))
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

X = np.column_stack((np.ones_like(x), x))
coefficients, residuals, rank, singular_values = np.linalg.lstsq(X, y, rcond=None)

a, b = coefficients
y_pred = X @ coefficients
sse = np.sum((y - y_pred) ** 2)

print("Матриця ознак X:\n", X)
print("a =", a)
print("b =", b)
print("Пряма: y ≈ a + b*x")
print("Прогнозні значення y:", y_pred)
print("Сума квадратів залишків:", sse)
````
